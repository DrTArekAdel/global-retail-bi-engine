```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import glob
```

```python

## Template to get all files in folder with its names then save them as dataframes 

# 1. Look inside your folder and grab every file path that ends with .csv
# (Change 'my_folder/' to the actual path where your files sit)
file_pattern = "E:\\power BI projects\\Maven Market\\data\\*.csv"
all_file_paths = glob.glob(file_pattern)

# 2. Run a loop to read every single CSV into a temporary list of DataFrames
df_list = [pd.read_csv(filename) for filename in all_file_paths]

for index, path in enumerate(all_file_paths):
    # 3. string splitting 
    file_name = path.split('\\')[-1]
    
    # 4. Pass the pre-calculated variable name into the f-string safely
    print(f"Index {index} is file: {file_name}")
```
